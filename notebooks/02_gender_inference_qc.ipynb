# Inference quality control

Inspect confidence thresholds, coverage, and human-audit agreement. Labels are uncertain estimates, not statements of identity.

```python
from pathlib import Path
import pandas as pd

path = Path('../data/processed/arxiv_authors_final.csv')
if path.exists():
    authors = pd.read_csv(path)
    threshold = 0.90
    display(authors['inferred_gender'].value_counts(dropna=False, normalize=True))
    display((authors['confidence'] >= threshold).mean())
else:
    print('Add processed author records to begin quality checks.')
```
