# Statistical modeling

Estimate subfield-level representation and uncertainty after validation. Avoid causal claims from descriptive observational data.

```python
from pathlib import Path
import pandas as pd

path = Path('../data/processed/arxiv_authors_final.csv')
if path.exists():
    authors = pd.read_csv(path)
    known = authors[authors['inferred_gender'] != 'unknown']
    summary = known.groupby('subfield')['inferred_gender'].value_counts(normalize=True)
    display(summary.rename('proportion').reset_index())
else:
    print('Add validated processed author records to begin modeling.')
```
